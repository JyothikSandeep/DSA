Fibanacci Series

```python
# Topic - 1D dp
#Recursion and memoization -> top down approach
# T(n) = T(n-1) + T(n-2)
# Time complexity : O(n)
# Space complexity : O(n) -> Memo dictionary + Recursion stack 
n=10

memo={}
def fibonacci(n):
    if(n==1):
        return 1
    if(n==0):
        return 0
    if(n in memo):
        return memo[n]
    memo[n]=fibonacci(n-1)+fibonacci(n-2)
    return memo[n]

print(fibonacci(10))


# tabulation -> bottom up approach
# Time complexity: O(n)
# Space complexity: O(n) -> Array space

dp=[-1]*(n+1)
dp[0]=0
dp[1]=1
for i in range(2,n+1):
    dp[i]=dp[i-1]+dp[i-2]
print(dp[n])


# Space optimization
# Time complexity : O(n)
# Space complexity : O(1)
n1,n2=0,1

for i in range(n-1):
    n1,n2=n2,n1+n2
print(n2)

```

Cimbing Stairs: leetcode 70:

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

Example 1:

Input: n = 2
Output: 2
Explanation: There are two ways to climb to the top.
1. 1 step + 1 step
2. 2 steps
Example 2:

Input: n = 3
Output: 3
Explanation: There are three ways to climb to the top.
1. 1 step + 1 step + 1 step
2. 1 step + 2 steps
3. 2 steps + 1 step

```python
# Topic : 1 D dp
# Climbing stairs:
# Time complexity : O(n)
#  Space Complexity : O(n)+recursive stack space
n=10
memo={}
def call(n):
    if(n<=0):
        if(n<0):
            return 0
        return 1
    if(n in memo):
        return memo[n]
    memo[n]=call(n-1)+call(n-2)
    return memo[n]
print(call(n))

# Tabulation 
# Time Complexity : O(n)
# Space Complexity : O(n)

dp=[-1]*(n+1)
dp[0]=1
dp[1]=1
for i in range(2,n+1):
    dp[i]=dp[i-1]+dp[i-2]
print(dp[-1])

# Tabulation:
# Time complexity O(N)
# Space complexity O(1)

dp1=1
dp2=1
for i in range(2,n+1):
    dp1,dp2=dp2,dp1+dp2
print(dp2)
```

Given a list of stones positions (in units) in sorted ascending order, determine if the frog can cross the river by landing on the last stone. Initially, the frog is on the first stone and assumes the first jump must be 1 unit.

If the frog's last jump was k units, its next jump must be either k - 1, k, or k + 1 units. The frog can only jump in the forward direction.

Example 1:

Input: stones = [0,1,3,5,6,8,12,17]
Output: true
Explanation: The frog can jump to the last stone by jumping 1 unit to the 2nd stone, then 2 units to the 3rd stone, then 2 units to the 4th stone, then 3 units to the 6th stone, 4 units to the 7th stone, and 5 units to the 8th stone.
Example 2:

Input: stones = [0,1,2,3,4,8,9,11]
Output: false
Explanation: There is no way to jump to the last stone as the gap between the 5th and 6th stone is too large.
 

```python
# Topic - 2d DP

# memoization:
# Time complexity:O(n^2)+ recursion stack pos → up to n stones jump → in worst case can go up to n
# space O(n^2)

class Solution:
    def canCross(self, stone: List[int]) -> bool:
        n=len(stone)
        stone_set=set(stone)
        memo={}
        target=stone[-1]
        def call(pos,jump):
            if((pos,jump) in memo):
                return memo[(pos,jump)]
            if(pos==target):
                return True
            if(pos<=0):
                return False
            for i in [0,1,-1]:
                current_pos=pos+jump+i
                if(jump+i>0 and current_pos in stone_set):
                    if(call(current_pos,jump+i)):
                        memo[(pos,jump)]=True
                        return True
            memo[(pos,jump)]=False
            return False 
        return call(1,1) if 1 in stone_set else False
    
    
# Tabulation:


class Solution:
    def canCross(self, stones: List[int]) -> bool:
        dp = {stone: set() for stone in stones}
        stone_set = set(stones)
        dp[0].add(0)
        for stone in stones:
            for jump in dp[stone]:
                for step in [jump, jump-1, jump+1]:
                    if step > 0 and stone + step in stone_set:
                        dp[stone + step].add(step)
            if len(dp[stones[-1]]) > 0:
                return True
        return False 

# Space optimization not possible because the jumps will always differ we can't predict.




            
            
        
```

You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed, the only constraint stopping you from robbing each of them is that adjacent houses have security systems connected and it will automatically contact the police if two adjacent houses were broken into on the same night.

Given an integer array nums representing the amount of money of each house, return the maximum amount of money you can rob tonight without alerting the police.

 

Example 1:

Input: nums = [1,2,3,1]
Output: 4
Explanation: Rob house 1 (money = 1) and then rob house 3 (money = 3).
Total amount you can rob = 1 + 3 = 4.
Example 2:

Input: nums = [2,7,9,3,1]
Output: 12
Explanation: Rob house 1 (money = 2), rob house 3 (money = 9) and rob house 5 (money = 1).
Total amount you can rob = 2 + 9 + 1 = 12.
 

```python
# 1d dynamic programming
#Time complexity:O(N)
# space complexity: O(n)+ recursive stack space

nums = [1,2,3,1]

memo={}
n=len(nums)
def call(n):
    if(n<0):
        return 0
    if(n in memo):  
        return memo[n]
    not_pick=call(n-1)
    pick=nums[n]+call(n-2)
    memo[n]=max(pick,not_pick)
    return memo[n]
print(call(n-1))

# tabulation:
# Time complexity: O(n)
# Space complexity:O(n)

n=len(nums)
dp=[-1]*(n)
dp1=nums[0]
if(n>1):
    dp[1]=max(nums[0],nums[1])

for i in range(2,len(nums)):
    pick=nums[i]+dp[i-2]
    not_pick=dp[i-1]
    dp[i]=max(pick,not_pick)
print(dp[n-1])





# Space optimization:

# Time complexity : O(n)
# Space complexity :O(1)

n=len(nums)
dp1=nums[0]
if(n==1):
    print(nums[0])
dp2=0
if(n>1):
    dp2=max(nums[0],nums[1])

for i in range(2,len(nums)):
    pick=nums[i]+dp1
    not_pick=dp2

    dp2,dp1=max(pick,not_pick),dp2
print(dp2)
```

62. Unique paths - 2D DP

There is a robot on an m x n grid. The robot is initially located at the top-left corner (i.e., grid[0][0]). The robot tries to move to the bottom-right corner (i.e., grid[m - 1][n - 1]). The robot can only move either down or right at any point in time.

Given the two integers m and n, return the number of possible unique paths that the robot can take to reach the bottom-right corner.

The test cases are generated so that the answer will be less than or equal to 2 * 109.

```python
# Time complexity :O(m*n)
# Space complexity: O(m*n)+ recursive stack space
memo={}
m=4
n=4
def call(m,n):
    if(m==0 and n==0):
        return 1
    if(m<0 or n<0):
        return 0
    if((m,n) in memo):
        return memo[(m,n)]
    memo[(m,n)]=call(m-1,n)+call(m,n-1)
    return memo[(m,n)]
call(m-1,n-1)

# Tabulation:
# Time complexity:O(M*N)
# Space complexity: O(m*n)

dp=[[0]*(n+1) for _ in range(m+1)]
dp[1][1]=1
for i in range(1,m+1):
    for j in range(1,n+1):
        if(i==1 and j==1):
            continue
        dp[i][j]=dp[i-1][j]+dp[i][j-1]
print(dp[m][n])


# Space optimization:

# Time complexity : O(m*n)
# Space complexity :O(n)

dp1=[0]*(n+1)
dp2=[0]*(n+1)
dp2[1]=1
for i in range(1,m+1):
    for j in range(1,n+1):
        if(i==1 and j==1):
            continue
        temp=dp1[j]+dp2[j-1]
        dp2[j]=temp
    dp1=dp2
    dp2=[0]*(n+1)
print(dp1[-1])
```

Return the number of possible unique paths that the robot can take to reach the bottom-right corner.

### Minimum Sum path - Leet code problem 64
Given a m x n grid filled with non-negative numbers, find a path from top left to bottom right, which minimizes the sum of all numbers along its path.

Note: You can only move either down or right at any point in time.

```python
# Memoization
# Time complexity: O(m*n)
# Space complexity: O(m*n)+ recursive_stack_space
grid = [[1,3,1],[1,5,1],[4,2,1]]
memo={}
rows=len(grid)
cols=len(grid[0])
def call(m,n):
    if(m==0 and n==0):
        return grid[0][0]
    if(m<0 or n<0):
        return float('inf')
    if((m,n) in memo):
        return memo[(m,n)]
    memo[(m,n)]=min(call(m-1,n),call(m,n-1))+grid[m][n]
    return memo[(m,n)]
print(call(rows-1,cols-1))

# Tabulation

# Time complexity: O(m*n)
# Space complexity : O(m*n)

rows=len(grid)
cols=len(grid[0])
dp=[[0]*(cols+1) for _ in range(rows+1)]
for i in range(cols+1):
    dp[0][i]=float('inf')
for i in range(rows+1):
    dp[i][0]=float('inf')
dp[1][1]=grid[0][0]
for i in range(1,rows+1):
    for j in range(1,cols+1):
        if(i==1 and j==1):
            continue
        dp[i][j]=min(dp[i-1][j],dp[i][j-1])+grid[i-1][j-1]
print(dp[rows][cols])


# Space optimization

# Time complexity: O(m*n)
# Space complexity : O(n)
rows=len(grid)
cols=len(grid[0])
dp1=[float('inf')]*(cols+1)
dp2=[0]*(cols+1)
dp2[1]=grid[0][0]
dp2[0]=float('inf')        
for i in range(1,rows+1):
    for j in range(1,cols+1):
        if(i==1 and j==1):
            continue
        dp2[j]=min(dp2[j-1],dp1[j])+grid[i-1][j-1]
    dp1=dp2
print(dp1[-1])
    
```

Given a triangle array, return the minimum path sum from top to bottom.

For each step, you may move to an adjacent number of the row below. More formally, if you are on index i on the current row, you may move to either index i or index i + 1 on the next row.

Example 1:

Input: triangle = [[2],[3,4],[6,5,7],[4,1,8,3]]
Output: 11
Explanation: The triangle looks like:
   2
  3 4
 6 5 7
4 1 8 3
The minimum path sum from top to bottom is 2 + 3 + 5 + 1 = 11 (underlined above).

```python
# Memoization-Bottom up
# Time complexity : O(N*M)
# Space complexity : O(N*M)+ recursive_stack_space
memo={}
triangle = [[2],[3,4],[6,5,7],[4,1,8,3]]
rows=len(triangle)
def call(m,n):
    if(m==rows-1):
        return triangle[m][n]
    if((m,n) in memo):
        return memo[(m,n)]
    memo[(m,n)]=min(call(m+1,n),call(m+1,n+1))+triangle[m][n]
    return memo[(m,n)]
print(call(0,0))

# Tabulation or space optimization:

dp = triangle[-1][:]   # copy last row
    
for i in range(len(triangle) - 2, -1, -1):
    for j in range(len(triangle[i])):
        dp[j] = triangle[i][j] + min(dp[j], dp[j+1])

print(dp[0])
```

931. Minimum Falling Path Sum
Given an n x n array of integers matrix, return the minimum sum of any falling path through matrix.

A falling path starts at any element in the first row and chooses the element in the next row that is either directly below or diagonally left/right. Specifically, the next element from position (row, col) will be (row + 1, col - 1), (row + 1, col), or (row + 1, col + 1).

Example 1:


Input: matrix = [[2,1,3],[6,5,4],[7,8,9]]
Output: 13
Explanation: There are two falling paths with a minimum sum as shown.

```python
# Memoization:

# Time complexity: O(m*n)
# Space complexity: O(m*n)
matrix = [[2,1,3],[6,5,4],[7,8,9]]
memo={}
rows=len(matrix)
cols=len(matrix[0])
def call(m,n):
    if(m==0 and n>=0 and n<cols):
        return matrix[m][n]
    if(m<0 or n<0 or m>=rows or n>=cols):
        return float('inf')
    if((m,n) in memo):
        return memo[(m,n)]
    memo[(m,n)]=min(call(m-1,n-1),call(m-1,n),call(m-1,n+1))+matrix[m][n]
    return memo[(m,n)]      
ans=[]
for i in range(cols):
    ans.append(call(rows-1,i))
print(min(ans))


# tabulation:

rows=len(matrix)
cols=len(matrix[0])
dp=[[0]*(cols+2) for _ in range(rows+1)]
dp[1]=[float('inf')]+matrix[0]+[float('inf')]
for i in range(cols+1):
    dp[0][i]=float('inf')
for i in range(rows+1):
    dp[i][0]=float('inf')
    dp[i][-1]=float('inf')
for i in range(2,rows+1):
    for j in range(1,cols+1):
        dp[i][j]=min(dp[i-1][j],dp[i-1][j+1],dp[i-1][j-1])+matrix[i-1][j-1]
print(min(dp[-1]))

# Space optimization:

rows, cols = len(matrix), len(matrix[0])
    
prev = matrix[0][:]

for i in range(1, rows):
    curr = [0] * cols
    for j in range(cols):
        up = prev[j]
        left = prev[j-1] if j > 0 else float('inf')
        right = prev[j+1] if j < cols-1 else float('inf')
        
        curr[j] = matrix[i][j] + min(up, left, right)
    
    prev = curr

print(min(prev))
```
